```python
#Definimos la neurona NOT
def neurona_NOT(x):
  #peso inhibitorio(negativo para voltear el resultado)
  w = -1

  #umbral
  umbral = 0

  #ahora multiplicamos la entrada por el peso
  suma = x * w

  #preguntamos¿llego o supero el umbral?
  if suma >= umbral:
    return 1       #si entro 0, se activa 1
  else:
    return 0       #si entro 1, se apaga 0


#probamos las entradas posibles
print("RESULTADOS DE LA NEURONA NOT:")
print("Entrada 0 = salida:", neurona_NOT(0))
print("Entrada 1 = salida:", neurona_NOT(1))

```

¿Como funciona la neurona?

La compuerta NOT funciona como un inversor
usando un peso inhibiitorio negativo de -1

°si entra un 0, l suma da 0, lo cual no alcanza el umbral
 y la neurona se activa en 1.

°si entra un 1, la suma da -1, quedando por debajo del umbral ,
 lo que apaga 0 la neurona.
