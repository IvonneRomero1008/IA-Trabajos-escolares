```python
#definimos la neurona OR
def neurona_OR(x1,x2):
  #pesos:que tanto le importa cada entrada(vale 1 cada una)
    w1= 1
    w2= 1

    #umbral: el minimo para activarse(con llegar a 1 basta)
    umbral = 1

    #sumamos la fuerza de las entradas
    suma = (x1 * w1) + (x2 * w2)

    #preguntamos:¿llego o supero el umbral?
    if suma >= umbral:
        return 1  # se enciende la neurona
    else:
        return 0  # se queda apagada


#probamos las combinaciones
print("RESULTADOS DE LA NEURONA OR:")
print("Entrada (0,0) = salida:",neurona_OR(0,0))
print("Entrada (0,1) = salida:",neurona_OR(0,1))
print("Entrada (1,0) = salida:",neurona_OR(1,0))
print("Entrada (1,1) = salida:",neurona_OR(1,1))
```

¿Como funciona esta neurona?
Esta neurona de McCullock funciona como si fuese un interruptor
con un umbral de 1.Recibiendo dos entradas x1 y x2, si al menos
una de las dos entradas envia 1, la suma llega al umbral y la neurona se activa, solo se queda apagada 0 cuando ambas entradas son 0
